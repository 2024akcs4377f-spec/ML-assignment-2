```python
import pandas as pd
import numpy as np
from sklearn.linear_model import Ridge
from sklearn.svm import SVR
from sklearn.metrics import mean_squared_error, r2_score, mean_absolute_error

# Load the datasets
x_train = pd.read_csv('/content/x_train.csv')
y_train = pd.read_csv('/content/y_train.csv')
x_test = pd.read_csv('/content/x_test.csv')
y_test = pd.read_csv('/content/y_test.csv')

# Flatten y arrays if they are loaded as DataFrames to avoid warnings
y_train_flat = y_train.values.ravel()
y_test_flat = y_test.values.ravel()

print("Datasets loaded successfully!")
print(f"Train shape: {x_train.shape}, Test shape: {x_test.shape}")
```

```python
# Train Ridge Regression Model
ridge_model = Ridge(alpha=1.0)
ridge_model.fit(x_train, y_train_flat)
ridge_preds = ridge_model.predict(x_test)

# Train SVM (Support Vector Regression) Model
svm_model = SVR(kernel='rbf')
svm_model.fit(x_train, y_train_flat)
svm_preds = svm_model.predict(x_test)

print("Models trained and predictions generated.")
```

```python
# Evaluate the models
metrics = {
    'Model': ['Ridge Regression', 'Support Vector Machine (SVR)'],
    'MAE': [
        mean_absolute_error(y_test_flat, ridge_preds),
        mean_absolute_error(y_test_flat, svm_preds)
    ],
    'MSE': [
        mean_squared_error(y_test_flat, ridge_preds),
        mean_squared_error(y_test_flat, svm_preds)
    ],
    'R2 Score': [
        r2_score(y_test_flat, ridge_preds),
        r2_score(y_test_flat, svm_preds)
    ]
}

# Display the comparison table
results_df = pd.DataFrame(metrics)
display(results_df)
```
